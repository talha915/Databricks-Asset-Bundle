```python
%sql
CREATE TABLE IF NOT EXISTS ai_lab_demo.system.execution_log
(
    run_id BIGINT,
    layer STRING,
    job_name STRING,
    status STRING,
    start_time TIMESTAMP,
    end_time TIMESTAMP,
    records BIGINT,
    error STRING
);
```

```python
%sql
CREATE TABLE IF NOT EXISTS ai_lab_demo.system.audit_log
(
    audit_id STRING,
    run_id BIGINT,
    event STRING,
    message STRING,
    timestamp TIMESTAMP
);
```

```python
%sql
CREATE TABLE IF NOT EXISTS ai_lab_demo.system.watermark
(
    pipeline_id BIGINT GENERATED ALWAYS AS IDENTITY,
    pipeline_name STRING,
    pipeline_src STRING,
    pipeline_destination STRING,
    ingestion_time TIMESTAMP,
    updated_at TIMESTAMP
);
```

```python
%sql
CREATE TABLE ai_lab_demo.system.quality_results
(
    pipeline_id BIGINT GENERATED ALWAYS AS IDENTITY,
    layer STRING,
    table_name STRING,
    check_name STRING,
    status STRING,
    failed_records BIGINT,
    message STRING,
    created_time TIMESTAMP
);
```

```python
%sql
CREATE TABLE ai_lab_demo.system.ai_quality_results
(
    id BIGINT GENERATED ALWAYS AS IDENTITY,
    table_name STRING,
    check_name STRING,
    input_record STRING,
    ai_result STRING,
    created_time TIMESTAMP
);
```
